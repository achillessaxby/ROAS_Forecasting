```python
import os
import pandas as pd
```

```python
output_folder = "updated_csv_with_elapsed=1_indexed"
os.makedirs(output_folder, exist_ok=True)

for file in os.listdir():
    if file.endswith(".csv"):
        file_path = os.path.join(os.getcwd(), file)
        df = pd.read_csv(file_path)

        if 'elapsed_months' in df.columns:
            df['elapsed_months'] += 1

            updated_file_path = os.path.join(output_folder, file)
            df.to_csv(updated_file_path, index=False)
            print(f"Updated file saved: {updated_file_path}")
        else:
            print(f"'elapsed_months' column not found in {file}")
```
