```python
import os
import pandas as pd
```

```python
files = [file for file in os.listdir() if file.endswith('.csv')]

dataframes = []

for file in files:
    print(f"Reading {file}...")
    df = pd.read_csv(file)
    dataframes.append(df)

if dataframes:
    concatenated_df = pd.concat(dataframes, ignore_index=True)
    concatenated_df.to_csv('M2_predictions.csv', index=False)
    print("All CSV files have been concatenated and saved'")
else:
    print("No CSV files found in the current directory.")
```
